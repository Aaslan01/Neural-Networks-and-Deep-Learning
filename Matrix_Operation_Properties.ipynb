```python
import numpy as np
```

```python
A = np.array([[1,2],[3,4]])
B = np.array([[5,6],[7,8]])
C = np.array([[1,5],[2,4]])
```

### Addition

**Cummulative**


```python
# A + B
np.add(A,B)
```

```python
# B + A
np.add(B,A)
```

**Associative**

```python
# ( A + B) + C
np.add(np.add(A,B), C)
```

```python
# A + (B + C)
np.add(A, np.add(B,C))
```

### Multiplication

**Associative**

```python
# (A * B) * C
np.matmul(np.matmul(A,B), C)
```

```python
# A * (B * C)
np.matmul(A, np.matmul(B,C))
```

**Distributive**

```python
# A * (B + C)
np.matmul(A, np.add(B,C))
```

```python
# A * B + A * C
np.add(np.matmul(A,B), np.matmul(A,C))
```

**Not Cummulative**

```python
# A * B
np.matmul(A,B)
```

```python
# B * A
np.matmul(B,A)
```
