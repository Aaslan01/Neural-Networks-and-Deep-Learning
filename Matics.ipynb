```python
A = [[1, 2], [3, 4]]
B = [[5, 6], [7, 8]]
C = [[1, 5], [2, 4]]
```

```python
# Addition Function
def add_matrices(X, Y):
    return [[X[i][j] + Y[i][j] for j in range(len(X[0]))] for i in range(len(X))]
```

```python
# Matric Multiplication
def multiply_matrices(X, Y):
    result = [[0 for _ in range(len(Y[0]))] for _ in range(len(X))]
    for i in range(len(X)):
        for j in range(len(Y[0])):
            for k in range(len(Y)):
                result[i][j] += X[i][k] * Y[k][j]
    return result
```

```python
# Lets start discusing Matric Properties:
# Commutative Property of Addition: A + B == B + A
print(add_matrices(A, B))
print(add_matrices(B, A))
```

```python
if add_matrices(A, B) == add_matrices(B, A):
    print("So it Proves Commutative Property of Addition: A + B == B + A")
```

```python
# Associative Property of Addition: (A + B) + C == A + (B + C)
print(add_matrices(add_matrices(A, B), C))
print(add_matrices(A, add_matrices(B, C)))
```

```python
if (add_matrices(add_matrices(A, B), C)) == (add_matrices(A, add_matrices(B, C))):
    print("So it Proves Associative Property of Addition: (A + B) + C == A + (B + C)")
```

```python
# Associative Property of Multiplication: (A * B) * C == A * (B * C)

print(multiply_matrices(multiply_matrices(A, B), C) == multiply_matrices(A, multiply_matrices(B, C)))  # True
```

```python
# Distributive Property: A * (B + C) == A * B + A * C
print(multiply_matrices(A, add_matrices(B, C)))
print(add_matrices(multiply_matrices(A, B), multiply_matrices(A, C)))
```

```python
if multiply_matrices(A, add_matrices(B, C)) == add_matrices(multiply_matrices(A, B), multiply_matrices(A, C)):
    print("So it Proves Distributive Property: A * (B + C) == A * B + A * C")
```

```python
# Non-Commutativity of Multiplication: A * B ≠ B * A
print(multiply_matrices(A, B))
print(multiply_matrices(B, A))
```

```python
if multiply_matrices(A, B) != multiply_matrices(B, A):
  print("So it Proves Non-Commutativity of Multiplication: A * B ≠ B * A")
```
